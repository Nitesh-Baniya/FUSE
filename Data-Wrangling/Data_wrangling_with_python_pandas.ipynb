## Data Wrangling

**Definition:**  
Data wrangling is the process of cleaning, transforming, and organizing raw data into a structured and usable format for analysis or modeling.

Raw data is almost never ready to use—columns might be mislabeled, values missing, formats inconsistent, etc. Wrangling is the work you do to fix all that so your analysis doesn’t give misleading results.

## Steps in Data Wrangling

1. Data collection / loading  
2. Data inspection (exploration)  
3. Handling missing data  
4. Data cleaning  
5. Data transformation  
6. Data integration (if needed)  
7. Data filtering and structuring  
8. Data validation / final check

```python
import pandas as pd

# Import the file and make a dataframe
reward_df = pd.read_csv('RewardsData.csv')

# View first 10 records
reward_df.head(10)
```

`isna()` in pandas is used to detect missing values in a dataset; it returns a boolean structure (DataFrame or Series) of the same shape where `True` indicates a missing (NaN) value and `False` indicates a non-missing value.

```python
reward_df.isna()
```

```python
# Since, True = 1 and False = 0, we can use sum() to get the count of total number of missing values.
missing_val_count = reward_df.isna().sum()
missing_val_count
```

```python
# Missing value shown in percentages
missing_val_percentage = missing_val_count / len(reward_df) * 100
missing_val_percentage
```

```python
# Here we can see that birthdate, city, and state have highest percentage of missing values. 
# But the last seen column have less percentage of missing values, so we can fill this column by using default values for further analysis.
reward_df['Last Seen']
```

```python
# We see that Last Seen column conatins the date time as string
# so we can convert this column into datetime format form the string type
reward_df['Last Seen'] = pd.to_datetime(reward_df['Last Seen'])
reward_df['Last Seen']
```

```python
reward_df['Last Seen'].max()
```

```python
reward_df['Last Seen'].min()
```

`fillna()` in pandas is used to replace missing (NaN) values in a dataset with a specified value or method (such as mean, median, or forward/backward fill).

```python
reward_df['Last Seen'].fillna(value=reward_df['Last Seen'].min())
```

```python
missing_val_count = reward_df.isna().sum()
missing_val_count
```

```python
reward_df_dropped = reward_df.dropna()
reward_df_dropped
```

```python
missing_val_count = reward_df_dropped.isna().sum()
missing_val_count
```

```python
# Drop the tags column from the new dataframe
reward_df_dropped = reward_df_dropped.drop(columns='Tags')
reward_df_dropped
```

```python
# Get list of unique values from city column
city_unique_values = reward_df_dropped["City"].unique().tolist()

# sort the values
# city_unique_values.()
city_unique_values.sort()
city_unique_values
```

```python
# We observe that there are variations for Winston Salem, we can replace all of them using replace()
reward_df_dropped['City'].replace(to_replace=['Winston Salem',  'Winston Salem ',  'Winston salem',  'Winston salem ', 'Winston-Salem', 'Winston-Salem ',
     'Winston-Salem, NC', 'Winston-salem'],value = "Winston-Salem")
```

```python
# Remove all the trailing and leading white spaces
reward_df_dropped['City'].str.strip()
```

```python
# Fix any casing issues 
reward_df_dropped['City'].str.title()
```

```python
# Get list of unique values from state column
state_unique_values = reward_df_dropped["State"].unique()

# sort the values
state_unique_values = sorted(state_unique_values)
state_unique_values
```

```python
# Replace each state with abbrebiation
for state_name in us_state_to_abbrev:
    reward_df_dropped['State'] = reward_df_dropped['State'].replace(
        state_name,
        us_state_to_abbrev[state_name]
    )
```

```python
# Pandas can do this directly using dictionary mapping
reward_df_dropped['State'] = reward_df_dropped['State'].replace(us_state_to_abbrev)
```

```python
# Get list of unique values from state column
state_unique_values = reward_df_dropped["State"].unique()

# sort the values
state_unique_values = sorted(state_unique_values)
state_unique_values
```

```python
# Convert the us_state_to_abbrev dictionary to df for merginin with reward df
states_df =pd.DataFrame(list(us_state_to_abbrev.items()), columns=['State Name', 'State']) 
# State column matches with the olumn of reward df while merginig
states_df.head(5)
```

```python
# Merging
reward_df_dropped = pd.merge(reward_df_dropped, states_df, on="State", how="left")
reward_df_dropped
```

```python
# Rename State column into State Abbrebiation
reward_df_dropped =  reward_df_dropped.rename(columns={"State":"State Abbrebiation"})
reward_df_dropped.head(5)
```

```python
#Move 'State Name' column to position after 'State Abbreviation' column, for this we use two methods together, pop() and insert()

# Step 1: Store the column data and remove it from its original position
state_name_col = reward_df_dropped.pop('State Name')

# Step 2: Insert the column back into the DataFrame at the new position
    # The first argument (3) specifies the new position for the column (zero-relative).
    # The second argument ('State Name') specifies the new inserted column name.
    # The third argument (col_a) is the data of the column.
reward_df_dropped.insert(4, "State Name", state_name_col)

reward_df_dropped.head(3)
```

```python
# Get list of unique values from zip column
zip_unique_values = reward_df_dropped["Zip"].unique()

# sort the values
zip_unique_values = sorted(zip_unique_values)
zip_unique_values
```

```python
# Truncate all the zip codes more than 5 digits into 5 digits at max, we use string slicing
reward_df_dropped['Zip'] = reward_df_dropped['Zip'].str[:5]
```

```python
# Learn Boolean indexing
```

Boolean indexing in pandas is a way to filter data in a DataFrame or Series using True/False conditions.

Instead of selecting rows by position or label, you create a condition (a boolean mask), and pandas returns only the rows where the condition is True.

```python
# create a boolean index to check for zip codes with a length of five digits
# Assign true if length == 5, else false
zip_code_boolean_index = reward_df_dropped['Zip'].str.len() == 5

# display result of boolean index
zip_code_boolean_index
```

```python
# Drop those zip codes not having length ewuals to 5
reward_df_dropped = reward_df_dropped[zip_code_boolean_index]
```

```python
# Get list of unique values from zip column
zip_unique_values = reward_df_dropped["Zip"].unique()

# sort the values
zip_unique_values = sorted(zip_unique_values)
zip_unique_values
```
